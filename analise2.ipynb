```python
import pandas as pd
import numpy as np
import openpyxl


# Ler uma guia específica
jaguarao  = pd.read_excel("data_RS.xlsx", sheet_name= "JAGUARAO")

bage  = pd.read_excel("data_RS.xlsx", sheet_name= "BAGE")

santana =  pd.read_excel("data_RS.xlsx", sheet_name= "SANTANA DO LIVRAMENTO ")

uruguaina = pd.read_excel("data_RS.xlsx", sheet_name= "URUGUAIANA")

saoBorja = pd.read_excel("data_RS.xlsx", sheet_name= "SAO BORJA")

portoAlegre =  pd.read_excel("data_RS.xlsx", sheet_name= "PORTO ALEGRE")

chui = pd.read_excel("data_RS.xlsx", sheet_name= "CHUI")

rioGrande = pd.read_excel("data_RS.xlsx", sheet_name= "RIO GRANDE")




```

```python
import pandas as pd
import matplotlib.pyplot as plt


# Agrupar por ano
df_ano = (
    jaguarao
    .groupby("Ano")["Peso_KG"]
    .sum()
    .reset_index()
)

# Converter para milhões
df_ano["Peso_milhoes"] = df_ano["Peso_KG"] / 1_000_000

# Criar gráfico
plt.figure()
plt.plot(df_ano["Ano"], df_ano["Peso_milhoes"], marker='o', label="Peso em milhões")

plt.title("Peso por Ano - Jaguarão")
plt.xticks(range(df_ano["Ano"].min(), df_ano["Ano"].max() + 1))
plt.ylabel("Peso (milhões de KG)")

# legenda
plt.legend()

plt.show()
```
